```python
import numpy as np
import matplotlib.pyplot as plt
from qiskit import transpile, QuantumCircuit
from qiskit_aer import AerSimulator
import qiskit.quantum_info as qi
from qiskit.visualization import plot_histogram
from qiskit_ibm_runtime import QiskitRuntimeService
from google.colab import drive
from qiskit_aer.noise import (
    NoiseModel,
    QuantumError,
    ReadoutError,
    depolarizing_error,
    pauli_error,
    thermal_relaxation_error,
)
drive.mount("/content/drive")
```

# Parametes

```python
Gamma = 0.25
delta_t = 0.5
g = np.sqrt(Gamma/delta_t)
omega = 1
N = int(30/delta_t)
nshots = 1024
```

# Load results from Linblad equation

```python
file_path = (
    '/content/drive/MyDrive/'
    'cont_solution_dt_0.5_g_0.71_omegas_1.txt'
)

data_cont = np.loadtxt(file_path, comments='#')

t_cont = data_cont[:, 0]
sx_cont = data_cont[:, 1]
sy_cont = data_cont[:, 2]
sz_cont = data_cont[:, 3]
```

# Refresh Ancilla

### Noise - flip

```python
def sz_2_circuit(simulator_aer_noise):
  sz_2_noiseflip1 = []
  for niter in range(0,N+1):
      qc = QuantumCircuit(2,1)
      qc.ry(np.pi-np.pi/3,0)
      qc.rz(-np.pi/3,0)
      for i in range(0,niter):
          qc.rxx(g*delta_t,0,1)
          qc.ryy(g*delta_t,0,1)
          qc.rz(-omega*delta_t,0)
          qc.reset(1)
      qc.measure(0,0)
      transpiled_qc = transpile(
      qc,
      backend=simulator_aer_noise,
      optimization_level=3
  )
      job = simulator_aer_noise.run(transpiled_qc,shots = nshots)
      results = job.result().get_counts()
      valor = (results.get('0',0) - ((results.get('1',0)))) / nshots
      sz_2_noiseflip1.append(-valor)
  return sz_2_noiseflip1
```

```python
def sx_2_circuit(simulator_aer_noise):
  sx_2_noiseflip1 = []
  for niter in range(0,N+1):
      qc = QuantumCircuit(2,1)
      qc.qubits
      qc.ry(np.pi-np.pi/3,0)
      qc.rz(-np.pi/3,0)
      for i in range(0,niter):
          qc.rxx(g*delta_t,0,1)
          qc.ryy(g*delta_t,0,1)
          qc.rz(-omega*delta_t,0)
          qc.reset(1)
      qc.h(0)
      qc.measure(0,0)

      transpiled_qc = transpile(
      qc,
      backend=simulator_aer_noise,
      optimization_level=3
  )
      job = simulator_aer_noise.run(transpiled_qc,shots = nshots)


      results = job.result().get_counts()
      valor = (results.get('0', 0) - (results.get('1', 0))) / nshots
      sx_2_noiseflip1.append(valor)
  return sx_2_noiseflip1
```

```python
def sy_2_circuit(simulator_aer_noise):
  sy_2_noiseflip1 = []
  for niter in range(0,N+1):
      qc = QuantumCircuit(2,1)
      qc.qubits
      qc.ry(np.pi-np.pi/3,0)
      qc.rz(-np.pi/3,0)
      for i in range(0,niter):
          qc.rxx(g*delta_t,0,1)
          qc.ryy(g*delta_t,0,1)
          qc.rz(-omega*delta_t,0)
          qc.reset(1)
      qc.sdg(0)
      qc.h(0)
      qc.measure(0,0)

      transpiled_qc = transpile(
      qc,
      backend=simulator_aer_noise,
      optimization_level=3
  )
      job = simulator_aer_noise.run(transpiled_qc,shots = nshots)

      results = job.result().get_counts()
      valor = (results.get('0', 0) - (results.get('1', 0))) / nshots
      sy_2_noiseflip1.append(-valor)
  return sy_2_noiseflip1
```

```python
simualtor_aer = AerSimulator()

time = np.arange(N+1) * delta_t



fig_x, ax_x = plt.subplots()
fig_y, ax_y = plt.subplots()
fig_z, ax_z = plt.subplots()


p_readout_list = [0,0.005,0.01, 0.05, 0.1]
prob = np.zeros((2,2))

for p_readout in p_readout_list:

  p10 = p_readout
  p00 = 1-p10
  p01 = p10
  p11 = 1-p01

  prob[0] = [p00,p10]
  prob[1] = [p01,p11]

  print(prob)

  # QuantumError objects
  error_readout = ReadoutError(prob, 1)

  noise_readout = NoiseModel()
  noise_readout.add_all_qubit_readout_error(error_readout)
  simulator_aer_noise = AerSimulator(noise_model=noise_readout)

  sx_2 = sx_2_circuit(simulator_aer_noise)
  sy_2 = sy_2_circuit(simulator_aer_noise)
  sz_2= sz_2_circuit(simulator_aer_noise)

  datos = np.column_stack((time, sx_2, sy_2, sz_2))

  np.savetxt(
  f"/content/drive/MyDrive/qiskit_noisereadout_sxsysz_ns1024_preadout_{p_readout}_sym.txt",
  datos,
  header="t sx sy sz",
  fmt="%.10f"
  )

  ax_x.plot(time, sx_2, label=rf'$p_{{10}}={p_readout}$')
  ax_y.plot(time, sy_2, label=rf'$p_{{10}}={p_readout}$')
  ax_z.plot(time, sz_2, label=rf'$p_{{10}}={p_readout}$')

ax_x.plot(
    t_cont, sx_cont,
    color='black', linestyle='--', linewidth=2,
    label='Continuous solution'
)

ax_y.plot(
    t_cont, sy_cont,
    color='black', linestyle='--', linewidth=2,
    label='Continuous solution'
)

ax_z.plot(
    t_cont, sz_cont,
    color='black', linestyle='--', linewidth=2,
    label='Continuous solution'
)
# Configurar figura de X
ax_x.set_xlabel('t')
ax_x.set_ylabel(r'$\langle \sigma_x \rangle$')
ax_x.set_title(r'Evolution of $\langle \sigma_x \rangle$ - Readout errors ($p_{01} = p_{10}$))')
ax_x.legend()

# Configurar figura de Y
ax_y.set_xlabel('t')
ax_y.set_ylabel(r'$\langle \sigma_y \rangle$')
ax_y.set_title(r'Evolution of $\langle \sigma_y \rangle$- Readout errors ($p_{01} = p_{10}$)')
ax_y.legend()

# Configurar figura de Z
ax_z.set_xlabel('t')
ax_z.set_ylabel(r'$\langle \sigma_z \rangle$')
ax_z.set_title(r'Evolution of $\langle \sigma_z \rangle$- Readout errors ($p_{01} = p_{10}$)')
ax_z.legend()

plt.show()
```
